### W9A1 
Viết hàm merge hai danh sách đã sắp xếp mà không dùng bất cứ thuật toán sắp xếp nào

```python
def merge(lst1,lst2):
    lst=lst1+lst2
    return lst
lst1=list(map(int,input().split()))
lst2=list(map(int,input().split()))
print(merge(lst1,lst2))
```
