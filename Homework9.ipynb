### W9A19
Loại bỏ khỏi danh sách các phần tử trùng lặp, giữ lại đúng 1 lần xuất hiện, in lại theo đúng thứ tự
xuất hiện


```python
lst=list(map(int,input().split()))
def danh_sach(lst):
    arr=[]
    for i in lst:
        if i not in arr:
            arr.append(i)
    return arr
print(*(danh_sach(lst)))
```

### W9A20
Thực hiện chuẩn hóa chuỗi kí tự:
a. bỏ khoảng trắng đầu cuối
b. các khoảng trắng giữa chỉ còn 1 dấu cách
c. viết hoa chữ cái đầu mỗi từ
KHÔNG DÙNG CÁC HÀM SẴN CỐ


```python
def hoa(c):
    if 'a'<=c<='z':
        return chr(ord(c)-32)
    return c
def thuong(c):
    if 'A'<=c<='Z':
        return chr(ord(c)+32)
    return c
def chuan_hoa(s):
    ket_qua=""
    viet_hoa=True
    for c in s:
        if c==' ':
            viet_hoa=True
        else:
            if viet_hoa:
                if len(ket_qua)>0:
                    ket_qua+=" "
                ket_qua+=hoa(c)
                viet_hoa=False
            else:
                ket_qua+=thuong(c)
    return ket_qua
s=input()
print(chuan_hoa(s))
        
```
