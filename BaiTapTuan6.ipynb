### W6A1
Nhập vào một dòng gồm các số nguyên. Hãy loại bỏ phần tử trùng nhưng
giữ nguyên thứ tự xuất hiện đầu tiên, rồi in ra list kết quả.  
Yêu cầu in ra đúng dạng list Python, ví dụ [1, 2, 3]

```python
n=list(map(int,input().split()))
ket_qua=[]
for x in n:
    if x not in ket_qua:
        ket_qua.append(x)
print(ket_qua)
```

### W6A2
Nhập một dòng số nguyên. In ra list tổng tích lũy (phần tử i là tổng từ đầu
tới vị trí i), giữ nguyên thứ tự

```python
def tong(lst):
    arr=[int(x) for x in lst.split()]
    ttl=0
    chung=[]
    for x in arr:
        ttl+=x
        chung.append(ttl)
    return chung
```

### W6A3
Nhập một tuple số nguyên (một dòng) và số nguyên k (dòng kế). In ra tuple
sau khi xoay trái k vị trí. (Giữ nguyên thứ tự tương đối sau xoay; không sắp
xếp.)

```python
tup=tuple(map(int,input().split()))
k=int(input())
def xoay(tup,k):
    lst=list(tup)
    arr1=[]
    arr2=[]
    for i in range (k):
        arr1.append(lst[i])
    for j in range (k,len(lst)):
        arr2.append(lst[j])
    return tuple(arr2+arr1)
print(xoay(tup,k))
    
```

### W6A5
Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ..., 'zeros':
...} (đếm số >0, <0, =0)

```python
lst=map(int,input().split())
def so(lst):
    dic={'positives': 0, 'negatives': 0, 'zeros': 0}
    for i in lst:
        if i>0:
            dic['positives']+=1
        elif i<0:
            dic['negatives']+=1
        else:
        	dic['zeros']+=1
    return dic
print(so(lst))
        
```
