### W8A1
[Cylinder]  
Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và chiều cao
(kiểu số thực) với hai phương thức:tính diện tích bề mặt (diện
tích toàn phần) và thể tích của hình trụ.  
Đầu vào  
Đâu vào từ bàn phím gồm một dòng duy nhất chứa bán kính và
chiều cao của hình trụ, cách nhau bởi một dấu cách.  
Đầu ra  
In ra màn hình diện tích bề mặt và thể tích của hình trụ (sử
dụng lại những hàm vừa viết, π=3.14).  
(Kết quả làm tròn đến 2 chữ số sau dấu phẩy).

```python
class Cylinder:
    def __init__(self,r,h):
        self.bankinh=r
        self.chieucao=h
    def getArea(self):
        return 2*3.14*self.bankinh**2 +2*3.14*self.bankinh*self.chieucao
    def getVolume(self):
        return self.chieucao*self.bankinh**2*3.14
    def __str__(self):
        return f"({self.bankinh},{self.chieucao})"
b,c=map(float,input().split())
cy1=Cylinder(b,c)
print(cy1)
print(f'Dien tich toan phan:{cy1.getArea()}')
print(f'The tich:{cy1.getVolume():.2f}')
        
```

### W8A2
[Date]  
Xây dựng lớp Date biểu diễn ngày tháng năm với hai phương
thức kiểm tra xem một ngày tháng nào đó có hợp lệ hay không
và phương thức tính ngày kế tiếp của ngày tháng năm nào đó
nếu nó là hợp lệ  
Nhập vào một ngày tháng năm, (ngày tháng năm biểu diễn
dưới dạng dd/mm/yyyy). Nếu ngày tháng năm không tồn tại, in
ra dòng chữ INVALID. Nếu là hợp lệ in ra ngày kế tiếp của nó.  
Đầu vào   
Đầu vào từ bàn phim chứa dữ liệu ngày, tháng, năm dưới dạng dd/mm/yyyy 
.  
Đầu ra  
In ra màn hình ngày kế tiếp của ngày vừa nhập dưới dạng
dd/mm/yyyy
.   Nếu ngày tháng không tồn tại, in ra INVALID

```python
class MyDate:
    def __init__(self,day,month,year):
        self.day=day
        self.month=month
        self.year=year
    def checkValid(self):
        maxday=-1
        match self.month:
            case 1|3|5|7|8|10|12:
                maxday=31
            case 4|6|9|11:
                maxday=30
            case 2:
                if (self.year%4==0 and self.year%100!=0) or (self.year%400==0):
                    maxday=29
                else:
                    maxday=28
        return 1<=self.day<=maxday and self.year>0
    def getNextDay(self):
        if (self.checkValid()):
            day1=MyDate(self.day+1,self.month,self.year)
            if day1.checkValid():
                return day1
            day1=MyDate(1,self.month+1,self.year)
            if day1.checkValid():
                return day1
            return MyDate(1,1,self.year+1)
    def __str__(self):
        return f'{self.day:02d}/{self.month:02d}/{self.year}'

d,m,y=map(int,input().split('/'))
day1=MyDate(d,m,y)
print(day1.getNextDay())
```

Đầu ra:  
Thông báo M có phải gốc tọa độ không hoặc có nằm trên trục
tung hoặc trục hoành không, khoảng cách từ M tới gốc tọa độ
là bao nhiêu (lấy 2 chữ số phần thập phân)  

```python
class Point:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def check(self):
        if self.x==self.y==0:
            return 'La goc toa do'
        elif self.x==0:
            return "Nam tren truc tung"
        elif self.y==0:
            return "Nam tren truc hoanh"
        else:
            return f'{((self.x**2+self.y**2)**0.5):.2f}'
x,y=map(int,input().split())
kq=Point(x,y)
print(kq.check())
```
