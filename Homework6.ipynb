### W6A16 
Nhập vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma
trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải

```python
n, m = map(int, input().split())

# Bước 2: Nhập mảng n hàng
matrix = []
for i in range(n):
    # Đọc một dòng, chuyển thành list số nguyên rồi thêm vào matrix
    row = list(map(int, input().split()))
    matrix.append(row)

# Bước 3: In ma trận theo định dạng
for row in matrix:
    for x in row:
        # In phần tử x chiếm 4 vị trí, căn phải, không xuống dòng
        print(f"{x:>4}", end="")
    # Sau khi in hết một hàng thì xuống dòng
    print()
```

### W6A20 
Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1

```python
def vi_tri(lst,k):
    for i in range (len(lst)):
        if lst[i] ==k:
            return i
    return -1
lst=list(map(int,input().split()))
k=int(input())
print(vi_tri(lst,k))
```

### W6A21
Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ
chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện

```python
lst=input().split()
def chuoi(lst):
    dic={}
    for i, tu in enumerate(lst):
        if tu not in dic:
            dic[tu]=[i]
        else:
            dic[tu].append(i)
    for tu in dic:
        dic[tu]=tuple(dic[tu])
    return dic

print (chuoi(lst))
        
```

### W6A22
Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
trong list

```python
def ln(n):
    if not n: 
        return None
    mx=n[0]
    vitri=0
    for i in range(len(n)):
        if n[i]>mx:
            mx=n[i]
            vitri=i
    return vitri
```

### W6A23 
VIết hàm xóa đi các số < x có trong mảng

### W6A24
Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí

trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9

```python
def doi(lst):
    day0=[]
    dayso=[]
    for i in range (len(lst)):
        if lst[i]==0:
            day0.append(lst[i])
        else:
            dayso.append(lst[i])
    return day0+dayso
```

### W6A26
Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2
phần tử đầu và cuối

```python
def tong(lst):
    tong=0
    for i in range(1,len(lst)-1):
        if lst[i]-lst[i-1]>0 and lst[i]-lst[i+1]>0:
            tong+=lst[i]
    return tong
```
